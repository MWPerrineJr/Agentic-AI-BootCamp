```python
# Load the API Key
from dotenv import load_dotenv
load_dotenv(override = True)

import os
openai_api_key = os.getenv("OPENAI_API_KEY")
groq_api_key = os.getenv("GROQ_API_KEY")
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")
deepseek_api_key = os.getenv("DEEPSEEK_API_KEY")

deepseek_base_url = 'https://api.deepseek.com'

if openai_api_key:
    print(f"OpenAI API Key loaded successfully and startswith {openai_api_key[:5]}")
else: 
    print("OpenAI Key not found")

if groq_api_key:
    print(f"Groq API Key loaded successfully and startswith {groq_api_key[:5]}")
else: 
    print("Groq API Key not found")

if openrouter_api_key:
    print(f"OpenRouter API Key loaded successfully and startswith {openrouter_api_key[:5]}")
else:
    print("OpenRouter API Key not found")

if deepseek_api_key:
    print(f"DeepSeek API Key loaded successfully and startswith {deepseek_api_key[:5]}")
else:
    print("DeepSeek API Key not found")
```

```python
# Start building the workflow
from openai import OpenAI
from IPython.display import Markdown, display


openai_python_client = OpenAI(api_key=deepseek_api_key,
                             base_url=deepseek_base_url)

response = openai_python_client.chat.completions.create(
    model='deepseek-chat',
    messages=[
        {
            'role': 'user',
            'content': ', What is 2 * 2'
        }
    ],
    stream=False
)

raw_response = response.choices[0].message.content
display(Markdown(raw_response))
```

```python
# Ask some questions
question = "Please propose a challenging question to assess someone's IQ. Respond only with the question."

messages=[
    {
        'role': 'user',
        'content': question
    },
]
```

```python
# Ask the question to the model
import openai
openai.api_key = os.getenv("OPENAI_API_KEY")
response = openai.chat.completions.create(
    model='gpt-6-luna',
    messages=messages,
    stream=False
)
question = response.choices[0].message.content
display(Markdown(question))
```
